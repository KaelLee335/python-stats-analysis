```python
import random
d1=df['姓名']
n1 = random.sample(population=list(d1), k=3)   # 无放回抽取10个数据,k为抽样次数
n2 = random.choices(population=d1, k=10)        # 有放回抽取10个数据

print(' 无放回：', n1,'\n','有放回：', n2)
```

```python
df.loc[df['姓名'].isin(n1),:]
```

生成随机数

```python
# 生成不同分布的随机数
import numpy.random as np

np.seed(15)                              # 设定随机数种子
r1=np.standard_normal(size=5)            # 标准正态分布
r2=np.normal(loc=50,scale=5,size=5)      # 均值(loc)为50、标准差(scale)为5的正态分布
r3=np.uniform(low=0, high=10, size=5)    # 0~10之间的均匀分布

print('# 标准正态分布:','\n',r1,'\n''# 值为50、标准差为5的正态分布:','\n',r2,
      '\n''# 0~10之间的均匀分布:','\n',r3)  # 打印结果
```

```python
df['统计学']=df['统计学']+r1
df
```
